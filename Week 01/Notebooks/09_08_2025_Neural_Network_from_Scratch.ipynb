```python
def initial_parameters():
  W1 = np.random.randn(64, 4) - np.sqrt(2./4)
  B1 = np.random.randn(64, 1) - np.sqrt(2./64)
  W2 = np.random.randn(64, 64) - np.sqrt(2./64)
  B2 = np.random.randn(64, 1) - np.sqrt(2./64)
  W3 = np.random.randn(3, 64) - np.sqrt(2./64)
  B3 = np.random.randn(3, 1) - np.sqrt(2./3)
  return W1, B1, W2, B2, W3, B3
```

```python
def ReLU(Z1):
  return np.maximum(0, Z1)
```

```python
def softmax(Z3):
  Z3 = np.array(Z3)
  Z3 = Z3 - np.max(Z3, axis=0)
  exponent = np.exp(Z3)
  return exponent / np.sum(exponent, axis=0)
```

```python
def forward_propagation(W1, B1, W2, B2, W3, B3, X):
  Z1 = np.dot(W1, X) + B1
  A1 = ReLU(Z1)
  Z2 = np.dot(W2, A1) + B2
  A2 = ReLU(Z2)
  Z3 = np.dot(W3, A2) + B3
  A3 = softmax(Z3)
  return Z1, A1, Z2, A2, Z3, A3
```

```python
def one_hot(Y):
  Y = Y.astype(int)
  one_hot_Y = np.zeros((Y.size, Y.max() + 1))
  one_hot_Y[np.arange(Y.size), Y] = 1
  one_hot_Y = one_hot_Y.T
  return one_hot_Y
```

```python
def derivative_ReLU(Z):
  return Z > 0
```

```python
def backward_propagation(W1, B1, W2, B2, W3, B3, Z1, A1, Z2, A2, Z3, A3, X, Y):
  m = Y.size
  one_hot_Y = one_hot(Y)
  dZ3 = A3 - one_hot_Y
  dW3 = 1 / m * np.dot(dZ3, A2.T)
  dB3 = 1 / m * np.sum(dZ3, axis=1, keepdims = True)
  dZ2 = np.dot(W3.T, dZ3) * derivative_ReLU(Z2)
  dW2 = 1 / m * np.dot(dZ2, A1.T)
  dB2 = 1 / m * np.sum(dZ2, axis=1, keepdims = True)
  dZ1 = np.dot(W2.T, dZ2) * derivative_ReLU(Z1)
  dW1 = 1 / m * np.dot(dZ1, X.T)
  dB1 = 1 / m * np.sum(dZ1, axis=1, keepdims = True)
  return dW1, dB1, dW2, dB2, dW3, dB3
```

```python
def update_parameters(W1, B1, W2, B2, W3, B3, dW1, dB1, dW2, dB2, dW3, dB3, alpha):
  W1 = W1 - alpha * dW1
  B1 = B1 - alpha * dB1
  W2 = W2 - alpha * dW2
  B2 = B2 - alpha * dB2
  W3 = W3 - alpha * dW3
  B3 = B3 - alpha * dB3
  return W1, B1, W2, B2, W3, B3
```

```python
def prediction(A3):
  return np.argmax(A3, axis = 0)
```

```python
def get_accuracy(predictions, Y):
  return np.sum(predictions == Y) / Y.size
```

```python
def gradient_descent(X, Y, alpha, iterations):
  W1, B1, W2, B2, W3, B3 = initial_parameters()
  for i in range(iterations):
    Z1, A1, Z2, A2, Z3, A3 = forward_propagation(W1, B1, W2, B2, W3, B3, X)
    dW1, dB1, dW2, dB2, dW3, dB3 = backward_propagation(W1, B1, W2, B2, W3, B3, Z1, A1, Z2, A2, Z3, A3, X, Y)
    W1, B1, W2, B2, W3, B3 = update_parameters(W1, B1, W2, B2, W3, B3, dW1, dB1, dW2, dB2, dW3, dB3, alpha)
    if i % 50 == 0:
      predictions = prediction(A3)
      print("Prediction", predictions)
      accuracy = get_accuracy(predictions, Y)
      print("Accuracy", accuracy)
  return W1, B1, W2, B2, W3, B3
```
